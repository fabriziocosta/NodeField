# Tune a ZINC model

**Purpose:** Tune a ZINC model.

**Before you start:** ZINC data, molecular dependencies, and a search configuration. Use the Python environment prepared by [setup](../setup.ipynb).

**Results:** Trial results, best model, and sample comparisons.

Run the cells in order, reviewing the configuration before starting the main work. Data stays under `notebooks/datasets`; models and outputs use the project’s artifact folders.


Load the campaign tools and locate the data and output folders.


```python
print('Load the campaign tools and locate the data and output folders.')
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
%load_ext autoreload
%autoreload 2

from IPython.display import display

from conditional_node_field_graph_generator.notebooks import configure_notebook

NOTEBOOK_CONTEXT = configure_notebook(require_nsppk=True, print_torch=True)
globals().update(NOTEBOOK_CONTEXT)

from abstractgraph_graphicalizer.chem import draw_molecules
from conditional_node_field_graph_generator.extensions.demo import (
    load_zinc_hyperparameter_optimization_config,
    run_zinc_hyperparameter_optimization,
    summarize_zinc_hyperparameter_results,
)
```

Read and display the search configuration before starting the campaign.


```python
print('Read and display the search configuration before starting the campaign.')
CONFIG_PATH = REPO_ROOT / 'notebooks' / 'configs' / 'zinc_molecule_hyperparameter_optimization.yaml'
config = load_zinc_hyperparameter_optimization_config(CONFIG_PATH)

print(f'Config: {CONFIG_PATH}')
display(config)
```

Run the configured hyperparameter search and write its results. This may take a long time.


```python
%%time
print('Run the configured hyperparameter search and write its results. This may take a long time.')
result = run_zinc_hyperparameter_optimization(config, notebook_context=NOTEBOOK_CONTEXT)

print(f"Artifacts: {result['artifact_root']}")
if result['results_csv_path'] is not None:
    print(f"Results CSV: {result['results_csv_path']}")
```

Compare completed trials and inspect the best result.


```python
print('Compare completed trials and inspect the best result.')
results_df = summarize_zinc_hyperparameter_results(result)
display(results_df)

best_row = result['best_row']
display(best_row)
```

Display molecules generated by the best trial.


```python
print('Display molecules generated by the best trial.')
best_samples = result['best_samples']
best_violations = result['best_violation_counts']

if best_samples:
    draw_molecules(
        best_samples,
        legends=[f'#violations={int(v)}' for v in best_violations],
    )
else:
    print('Best trial returned no generated molecules.')
```
