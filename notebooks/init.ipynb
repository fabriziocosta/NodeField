# Notebook Init

Run this notebook once per kernel before working with the demo notebooks.


```python
import os
import subprocess
import sys
from pathlib import Path

repo_root = next(
    root for root in [Path.cwd(), *Path.cwd().parents]
    if (root / "conditional_node_field_graph_generator").exists()
)
nsppk_candidates = [
    repo_root.parent / "NSPPK",
    repo_root.parent / "nsppk",
]
local_nsppk_repo = next(
    (
        candidate.resolve()
        for candidate in nsppk_candidates
        if (candidate / "pyproject.toml").exists() or (candidate / "setup.py").exists()
    ),
    None,
)
if local_nsppk_repo is not None:
    print(f"Installing local NSPPK checkout from {local_nsppk_repo}")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-e", str(local_nsppk_repo)])
elif os.environ.get("NODEFIELD_SKIP_NSPPK") != "1":
    raise RuntimeError(
        "Could not find a local NSPPK checkout. Expected one of: "
        + ", ".join(str(path) for path in nsppk_candidates)
        + ". Set NODEFIELD_SKIP_NSPPK=1 only if the current notebook does not require NSPPK."
    )

ecosystem_root = repo_root.parent / "abstractgraph-ecosystem" / "repos"
ecosystem_packages = [
    ecosystem_root / "abstractgraph",
    ecosystem_root / "abstractgraph-ml",
    ecosystem_root / "abstractgraph-graphicalizer",
]
missing_ecosystem_packages = [path for path in ecosystem_packages if not (path / "pyproject.toml").exists()]
if missing_ecosystem_packages:
    raise RuntimeError(
        "Could not find local AbstractGraph ecosystem checkouts: "
        + ", ".join(str(path) for path in missing_ecosystem_packages)
    )

print(f"Installing local AbstractGraph ecosystem from {ecosystem_root}")
subprocess.check_call([sys.executable, "-m", "pip", "install", "-e", str(ecosystem_packages[0])])
subprocess.check_call([sys.executable, "-m", "pip", "install", "-e", str(ecosystem_packages[1])])
subprocess.check_call([sys.executable, "-m", "pip", "install", "-e", ".[chem]"], cwd=ecosystem_packages[2])

print(f"Installing NodeField notebook dependencies from {repo_root}")
subprocess.check_call([sys.executable, "-m", "pip", "install", "-e", "."], cwd=repo_root)
```
